```python
"""
EDA to examine and understand data.
"""
import pandas as pd
from pathlib import Path

data_directory = Path("./raw/")
```

```python
corpus = pd.read_parquet(data_directory / "corpus.parquet")

# Check null values
print(corpus.isnull().any())
corpus.head()
```

```python
queries = pd.read_parquet(data_directory / "queries.parquet")

print(queries.isnull().any())
```

```python
qrels = pd.read_parquet(data_directory / "qrels.parquet")

print(qrels.isnull().any())
qrels.head()
```

```python
# Examine an example query-id and associated corpus-id to see what a query and relevant docs look like
print(f"Query: {queries.loc[queries["_id"] == "8", "text"].iloc[0]}" )
print(f"Documents:")
print(f"{corpus.loc[corpus["_id"] == "566392", "text"].iloc[0]}")
print(f"{corpus.loc[corpus["_id"] == "65404", "text"].iloc[0]}")
```
